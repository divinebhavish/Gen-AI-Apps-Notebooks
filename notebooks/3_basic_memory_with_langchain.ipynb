```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
```

### Providing Converstation History

```python
prompts = [
    {"role":"user", "content":"Hello! My name is Bhavishya Mewara"},
    {"role":"ai", "content":"Hello Bhavishya! It's nice to meet you. How can I help you today??"},
    {"role":"user", "content":"What is my name?"}
]

res = llm.invoke(prompts)
```

```python
res.content[0]["text"]
```

### Context Memory Automation

```python
history = [] # [{"role":"user", "content":query}]

while True:
    query = input("user: ")
    if query.lower() in ["exit", "break", "bye", "quit"]:
        print("Exiting the chat. Goodbye!")
        break
    
    history.append({"role":"user", "content":query})
    print("User: ", query)
    
    res = llm.invoke(history)
    
    ans = res.content[0]["text"]
    
    history.append({"role":"ai", "content":ans})
    print("AI: ", ans, "\n")
    
```
