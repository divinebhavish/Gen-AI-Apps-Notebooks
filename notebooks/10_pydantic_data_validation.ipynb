### Now with Pydantic

```python
from pydantic import BaseModel, Field
from typing import List, Annotated
```

```python
class User(BaseModel):
    name : str
    age : int
```

```python
user = User(
    name = "Bhavish",
    age = 22
)

user
```

### Field Validation

```python
class Product(BaseModel):
    title: str = Field(max_length=10)
    des: str = Field(min_length=20)
    price: int = Field(gt=10, le=100)
```

```python
product1 = Product(
    title = "iPhone Duo",
    des = "This is amazing mobile from Apple",
    price = 80
)
product2 = Product(
    title = "SamsungS23",
    des = "This is amazing mobile from Samsung",
    price = 60
)
```

```python
class Order(BaseModel):
    id: str = Field(min_length=6, max_length=12)
    product:Product
```

```python
order = Order(id = "abcd12", product=product)
```

```python
order
```

```python
order.product.title
```

```python
class Carts(BaseModel):
    id: str = Field(min_length=6, max_length=12)
    items: List[Product]
    comments: str = Field(
        default = "No Comments"
    )
```

```python
cart = Carts(
    id = "abcd123",
    items = [product1, product2],
    comments = "This is my cart"
)
```

```python
cart
```

```python
for product in cart.items:
    print(product.title)
```

### Annotated

```python
commanValidator = Annotated[List[Product], Field(default=[])]
```

```python
class Carts(BaseModel):
    id: str = Field(min_length=6, max_length=12)
    items: commanValidator
    comments: str = Field(
        default = "No Comments"
    )
```

```python
class Order(BaseModel):
    products: commanValidator
```

```python
ord = Order(products=[product1, product2])
# ord = Order()
```
