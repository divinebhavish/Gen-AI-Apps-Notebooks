```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_community.utilities import GoogleSerperAPIWrapper
from langchain_groq import ChatGroq
from langchain.agents import create_agent
import textwrap
from langgraph.checkpoint.memory import MemorySaver
```

```python
llm = ChatGroq(model="openai/gpt-oss-20b")
search = GoogleSerperAPIWrapper()
```

```python
result = search.run("When Drishyam 3 ajay devgan's is releasing")

print(textwrap.fill(result, width=125))
```

### Agent Creation

```python
agent = create_agent(
    model = llm,
    tools = [search.run],
    system_prompt = "You are an agent and can search any ques on google",
    checkpointer = MemorySaver()
)
```

```python
question = "yes"
```

##### See we use memry saver langgraph model to make history and context memory so we use configurable which contain the the id in in which memory is stored

```python
res = agent.invoke( 
    {"messages":[{"role":"user", "content":question}]},
    {"configurable": {"thread_id": "asd12"}}
)
```

```python
print(res["messages"][-1].content)
```
