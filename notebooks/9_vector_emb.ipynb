#### step1 - Generating Vector embeddings

```python
documents = [
    "Machine learning is a subset of artificial intelligence",
    "Deep learning uses neural networks with multiple layers",
    "Python is a programming language popular for data science",
    "India won the cricket world cup in 2011",
    "AI models can understand and generate human language"
]
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings

```

```python
embeddings = GoogleGenerativeAIEmbeddings(
    model="gemini-embedding-001"
)
```

```python
vector = embeddings.embed_documents(documents)
```

```python
# print(vector)
len(vector[0])
```

#### step2 - Store vector embeddings into vector DB

```python
from langchain_chroma import Chroma
```

```python
# Because the above docuemts is in the from of ist fo string whcih is in text

vector_store = Chroma.from_texts(
    texts = documents,
    embedding = embeddings,
    persist_directory = "./vector_db"
)
```

```python
# k - here k is no. of most similar result we want 

query = "Data Science"
result = vector_store.similarity_search(query=query, k=2)
result
```
